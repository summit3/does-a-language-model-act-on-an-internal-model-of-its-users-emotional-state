# Phase 3: probes on cached activations

All numbers come from `scripts/03_probes.py` (laptop, no model). This notebook only loads the CSVs and displays the saved figures. Tables: `results/phase3_probe_by_layer.csv`, `phase3_baselines.csv`, `phase3_headline.csv`, `phase3_geometry_layer<L>.csv`, `phase3_pdist_bestlayer.csv`. Directions for Phase 4: `results/directions_layer<L>.pt`.
